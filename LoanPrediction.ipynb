### 4. Data Cleaning
- Here, one more thing is, Loan_ID is NOT a categorical feature for modeling.
- Even though it is “text”, it is an Identifier column (ID column) and has no predictive meaning.
#### > Why do we drop Loan_ID?
- Because, It is unique for every row, Machine learning learns patterns and IDs create noise, not patterns.
- ALWAYS DROP THE IDENTIFIERS FIRST.

```python
df. drop('Loan_ID', axis=1, inplace = True)
```

```python
df.info()
```

```python
df.isnull().sum()
```

### 4.1 Missing Value Treatment
### 4.1.1. Replacing Categorical Data Missing Values with mode 

- Here now if we observe, we can see that, self_ employed column has highest missing values of 32.
- since it is an important column and is a categorical column, we need to use the   

```python
#To replace the missing values in Self_ employed column, we do:
df['Self_Employed'] = df['Self_Employed'].fillna(df['Self_Employed'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
```

```python
df['Dependents'] = df['Dependents'].fillna(df['Dependents'].mode()[0])
```

```python
df['Married'] = df['Married'].fillna(df['Married'].mode()[0])
```

```python
df['Credit_History'] = df['Credit_History'].fillna(df['Credit_History'].mode()[0])
```

```python
df.isnull().sum()
```

### 4.1.2. Replacing Numerical Data Missing Values with median 

```python
df['LoanAmount'] = df['LoanAmount'].fillna(df['LoanAmount'].median())
```

```python
df['Loan_Amount_Term'] = df['Loan_Amount_Term'].fillna(df['Loan_Amount_Term'].median()) 
```

```python
df.isnull().sum()
```

### 5. Feature Engineering

#### Feature 1: Total Income

```python
df['Total_Income'] = df['ApplicantIncome']+ df['CoapplicantIncome']
df
```

#### Feature 2: Loan Burden

```python
df['Loan_to_Income'] = df['LoanAmount']/df['Total_Income']
df
```

### 6. Encoding
Machines cannot understand text, so convering the columns which has binary values to only numbers

#### 6.1. For Binary Columns:

```python
df['Gender'] = df['Gender'].map({'Male':1, 'Female':0})

df['Married'] = df['Married'].map({'Yes':1, 'No':0})

df['Self_Employed'] = df['Self_Employed'].map({'Yes':1, 'No':0})

df['Education'] = df['Education'].map({'Graduate':1, 'Not Graduate':0})

df['Loan_Status'] = df['Loan_Status'].map({'Y':1, 'N':0})
```

```python
df['Dependents'] = df['Dependents'].replace('3+', 3)
df['Dependents'] = df['Dependents'].astype(int)
```

```python
df['Property_Area'] = df['Property_Area'].map({
    'Rural':0,
    'Semiurban':1,
    'Urban':2
})
```

```python
df.head()
```

```python
df['Credit_History'] = df['Credit_History'].astype(int)
```

### 6. Scaling
